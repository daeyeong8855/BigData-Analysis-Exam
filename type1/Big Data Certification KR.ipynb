```python
import pandas as pd
import kagglehub

# Download latest version
path = kagglehub.dataset_download("agileteam/bigdatacertificationkr")
Customer_Data = pd.read_csv(path + "/Customer_Data.csv")
Titanic = pd.read_csv(path + "/Titanic.csv")
```

#### 데이터에서 IQR을 활용해 Fare컬럼의 이상치를 찾고, 이상치 데이터의 여성 수를 구하시오  
Data = Titanic

```python
print(Titanic['Fare'])

Fare = Titanic['Fare']

Q1 = Fare.quantile(0.25)
Q3 = Fare.quantile(0.75)

IQR = Q3-Q1
(Q1-1.5*IQR)
(Q3+1.5*IQR)

Fare['filter']=Fare[(Fare < (Q1-1.5*IQR)) or (Fare > (Q3+1.5*IQR))]
```
